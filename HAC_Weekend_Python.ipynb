```python
# data types in python
# data : textual or numerical or alphanumerical

# numerical : int(whole numbers : 3,4,5) and float (decimal numbers: 432.413)
# textual : string : str
# you need to define str data in " " or ' '
# name = "Chirantan" / name = 'Chirantan'

# type() : data type of that variable

price = 100
print(price)
print(type(price))
print("\n") # line break

temperature = 30.4
print(temperature)
print(type(temperature))
print("\n") # line break
city = "london"
print(city)
print(type(city))
print("\n") # line break
# whatever goes inside quotes, will be treated as string or textual data
password = "1234"
print(password)
print(type(password))
```

```python
# input function
# input()
# it accepts the input value from the user
# any input will be by default treated as string or textual data

number = input("Enter the number here : ")
print(number)

result = number * 2 # repeated the number twice
print(result)
print(type(result))
```

```python
# int(input()) : wrap inside int()
# int() will convert string data into integer format

number = int(input("Enter the number here : "))
print(number)

result = number * 2 # repeated the number twice
print(result)
print(type(result))
```
